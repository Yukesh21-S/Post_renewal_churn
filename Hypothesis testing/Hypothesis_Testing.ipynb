# Hypothesis

### H0: Tenure has no effect on churn
### H1: Lower tenure customers churn more

```python
# Null Hypothesis (H0):
# There is no difference in tenure between churned and retained customers.

# Alternative Hypothesis (H1):
# There is a significant difference in tenure between churned and retained customers.

# Test Used:
# Independent t-test

# Result:
# The p-value obtained is 3.80 × 10⁻¹⁵⁸, which is far less than the significance level of 0.05.

# Conclusion:
# The null hypothesis is rejected. This indicates that tenure has a significant impact on churn. Customers with lower tenure are more likely to churn.

from scipy.stats import ttest_ind

# Split data
x = df_selected[df_selected['prospect_outcome'] == 1]['Tenure_Years'].dropna()
y = df_selected[df_selected['prospect_outcome'] == 0]['Tenure_Years'].dropna()

stat, p = ttest_ind(x, y, equal_var=False)

print("Tenure vs Churn p-value:", p)

if p < 0.05:
    print("Reject H0 → Tenure affects churn\n")
else:
    print("Fail to reject H0\n")
print("Churn count:", len(x))
print("Won count:", len(y))

print("Unique churn values:", x.nunique())
print("Unique won values:", y.nunique())

print("Sample churn values:\n", x.head())
print("Sample won values:\n", y.head())
```

## TOTAL AMOUNT vs CHURN (t-test)

## CALL DIRECTION vs CHURN (Chi-square)
### H0: Call direction is independent of churn
### H1: Call direction affects churn

```python
cont_table = pd.crosstab(df_selected['Call_Direction'], df_selected['prospect_outcome'])

chi2, p, dof, expected = chi2_contingency(cont_table)

print("Call Direction vs Churn p-value:", p)

if p < 0.05:
    print("Reject H0 → Call direction affects churn")
else:
    print("Fail to reject H0")
```
