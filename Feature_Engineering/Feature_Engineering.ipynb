```python
import pandas as pd
```

```python
df_feature=spark.read.table("post_renewal_churn.cleaned_dataset.features_table").toPandas()
```

### Score per tenure
### New customer + high score ≠ same as old customer + high score
### Normalizes behavior

```python
df_feature['score_per_tenure'] = (
    df_feature['total_renewal_score_new'] / (df_feature['Tenure_Years'] + 1)
)
```

### High / Low tenure flag low tenure = high churn

```python
df_feature['is_low_tenure'] = (df_feature['Tenure_Years'] <= 3).astype(int)
```

### Score category

```python
df_feature['score_category'] = pd.cut(
    df_feature['total_renewal_score_new'],
    bins=[0, 30, 50, 100],
    labels=['Low', 'Medium', 'High']
)
```
