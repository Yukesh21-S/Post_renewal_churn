## **EDA Perform**

1. The dataset shows moderate class imbalance with retained customers significantly higher than churned customers.

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='prospect_outcome', data=pdf)
plt.title("Distribution of Customer Churn")
plt.show()
```

Customers with lower tenure exhibit a higher tendency to churn, indicating that long-term customers are more likely to renew.

```python
sns.boxplot(x='prospect_outcome', y='Tenure_Years', data=pdf)
plt.title("Tenure vs Churn")
plt.show()
```

Auto-renewal score shows minimal variation between churned and retained customers, suggesting weak predictive influence.

```python
sns.boxplot(x='prospect_outcome', y='auto_renewal_score', data=pdf)
plt.show()
```

```python
import pandas as pd

display(pd.crosstab(
    pdf['Serious_Complaint'],
    pdf['prospect_outcome'],
    normalize='index'
) * 100)
```

5. Customers with unknown payment methods demonstrate significantly higher churn compared to customers using card or BACS payment methods.

```python
sns.countplot(x='Payment_Method', hue='prospect_outcome', data=pdf)
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(x='Explicit_Competitor_Mention', hue='prospect_outcome', data=pdf)
plt.show()
```

Perform Univariate Analysis
Numerical Graphs

```python
import matplotlib.pyplot as plt
import seaborn as sns

num_cols = pdf.select_dtypes(include=['int64', 'float64', 'float32', 'int32']).columns

for col in num_cols:
    plt.figure(figsize=(5,3))
    sns.histplot(pdf[col], kde=True)
    plt.title(col)
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

cat_cols = pdf.select_dtypes(include=['object', 'category']).columns.tolist()

for col in cat_cols:
    plt.figure(figsize=(7,3))
    sns.countplot(x=pdf[col])
    plt.xticks(rotation=90)
    plt.title(col)
    plt.show()
```
